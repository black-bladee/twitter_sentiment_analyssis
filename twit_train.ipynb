```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import BernoulliNB
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score, classification_report
```

```python
df = pd.read_csv('twit_data.csv', encoding='latin-1', header=None)
df = df[[0, 5]]
df.columns = ['polarity', 'text']
print(df.head())
```

```python
df['polarity'] = df['polarity'].replace({4: 1})
```

```python
df
```

```python
def clean(text):
    return text.lower()

df['clean_text'] = df['text'].apply(clean)   
```

```python
print(df[['text', 'clean_text']].head())
```

```python
x_train,x_test,y_train,y_test = train_test_split(df['clean_text'],df['polarity'],random_state=42)
```

```python
vectorizer = TfidfVectorizer(max_features=5000, ngram_range= (1,2))
x_train_tfid = vectorizer.fit_transform(x_train)
x_test_tfid = vectorizer.fit_transform(x_test)
```

```python
log = LogisticRegression(max_iter=100)
log.fit(x_train_tfid,y_train)

pecdiction = log.predict(x_test_tfid)

print("acuraccy = ", accuracy_score(y_test,pecdiction))
```

```python
bnb = BernoulliNB()
bnb.fit(x_train_tfid, y_train)

bnb_pred = bnb.predict(x_test_tfid)

print("Bernoulli Naive Bayes Accuracy:", accuracy_score(y_test, bnb_pred))
print("\nBernoulliNB Classification Report:\n", classification_report(y_test, bnb_pred))
```
